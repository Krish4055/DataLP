```python
class animal():
    def __init__(self,name,age):
        self.name = name
        self.age = age

dog = animal("bud",3)
print(dog)
print(dog.name)
print(dog.age)
```

```python
dog2 = animal("Lucy",2)
print(dog2)
print(dog2.name)
print(dog2.age)
```

```python
### define a class with instance 

class animal():
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof")

dog4 = animal("bud",3)
dog4.bark()

```
