```python
try:
    num = int(input("Enter a Value"))
    result = 10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Denominator Can't be Zero")
except Exception as ex:
    print(ex)
```

```python
##try,except,block

try:
    num = int(input("Enter a Value"))
    result = 10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Denominator Can't be Zero")
except Exception as ex:
    print(ex)
else:
    print(f"This is the {result}")
```

```python
##try,except,block, finally

try:
    num = int(input("Enter a Value"))
    result = 10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Denominator Can't be Zero")
except Exception as ex:
    print(ex)
else:
    print(f"This is the {result}")
finally:
    print("Execution complete!")
```
