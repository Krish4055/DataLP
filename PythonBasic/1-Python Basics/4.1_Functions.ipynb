```python
name = input("Enter your Name : ")
def greet(name):
    print(f"Heyy {name} Good Morning !! ")

greet(name)
```

```python
# Positional Arguments 

def print_numbers(*args):
    for number in args :
        print(number)


print_numbers(1,2,3,4)
    
```

```python
# Key Word Arguments 

def print_details(**kwargs):
    for key,value in kwargs.items() :
        print(f"{key}:{value}")


print_details(name="Krish",last_name="Bhinmala")
    
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional Arguments:{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")


print_details(1,2,3,4,name="Krish",last_name="Bhinmala")
```
