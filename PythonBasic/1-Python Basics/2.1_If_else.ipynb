```python
# If Else
age = int(input("Enter Your Age"))

if age <= 13:
    print("You are a kid !")
elif age < 18:
    print("You are a Teenager!")
elif age >= 18:
    print("You are a Adult!")
else:
    print("Invalid Input!!!")
```

```python
# Nested If Else

number = int(input("Enter the Number:"))

if number > 0:
    print("The Number is a Positive number !")
    if number % 2 == 0 :
        print("The Number is a Even Number !")
    else:
        print("The Number is a Odd Number !")

else:
    print("The Number is Zero or a Negative Number")
```

```python
### Year is a Leap Year or Not 

Year = int(input("Enter The Year !!!"))

if (Year%4 == 0):
    if (Year%100 == 0):
        if (Year%400 == 0):
            print(Year,": The Year is a Leap Year!")
        else:
            print(Year,": The Year is not a Leap Year")
    else:
            print(Year,": The Year is a Leap Year")
else:
            print(Year,": The Year is not a Leap Year")
            
    
```
