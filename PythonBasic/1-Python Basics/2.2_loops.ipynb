```python
#Calculate The sum of n Natural Number using while loop

n = int(input("Enter the Number :"))

sum = 0

count = 1

while count <= n:
    sum = sum + count
    count = count + 1
print("The Sum of n Natural is :",sum)
```

```python
#Calculate The sum of n Natural Number using for loop

n = int(input("Enter the Number :"))

sum = 0


for i in range (n+1):
    sum = sum + i
print(sum)
```

```python
### Prime Number Between 1 to 100

for num in range (1,101):
    if num > 1:
        for i in range (2,num):
            if num % i == 0:
                break
        else:
            print(num)
```

```python
### Prime Number fo n Number
n = int(input("Enter number :"))

for num in range (1,n + 1):
    if num > 1:
        for i in range (2,num):
            if num % i == 0:
                break
        else:
            print(num)
```
